```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

input_dir = "/kaggle/input"
csv_files = []
for root, dirs, files in os.walk(input_dir):
    for f in files:
        if f.endswith(".csv"):
            csv_files.append(os.path.join(root, f))

print(f"Found {len(csv_files)} CSV file(s)")
for f in csv_files[:5]:
    print(" ", f)
```

```python
from tqdm.notebook import tqdm 
dfs = []

CORE_COLS = [
    "date", "serial_number", "model", "capacity_bytes", "failure",
    # The SMART attributes most predictive of failure (well-known from research)
    "smart_5_raw",    # Reallocated sectors
    "smart_9_raw",    # Power-on hours
    "smart_187_raw",  # Uncorrectable errors
    "smart_188_raw",  # Command timeout
    "smart_197_raw",  # Pending sectors
    "smart_198_raw",  # Uncorrectable sectors
]
for f in tqdm(csv_files):
    dfs.append(pd.read_csv(f, usecols=lambda c: c in CORE_COLS, low_memory=False))

df = pd.concat(dfs, ignore_index=True)
```

```python
N = 7   # "will this drive fail within the next 7 days?"


death_date = df["date"].where(df["failure"] == 1)

# Copy that date onto ALL rows of the same drive, so day 1 of a drive knows
# when its drive will eventually die.
df["fail_date"] = death_date.groupby(df["serial_number"], observed=True).transform("max")

# How many days until death, for each row (4 = dies in 4 days, 0 = dies today)
df["days_to_fail"] = (df["fail_date"] - df["date"]).dt.days

# target = 1 only if the drive dies within the next N days (today included)
df["target"] = ((df["days_to_fail"] >= 0) & (df["days_to_fail"] < N)).astype("int8")

# Remove the last N days of the dataset. A drive that dies right after the
# data ends would look healthy there, which gives wrong labels.
df = df[df["date"] <= df["date"].max() - pd.Timedelta(days=N)]

print("positives:", df["target"].sum(), " (at most", 1522 * N, ")")

# ---------- 2. SPLIT BY TIME ----------
# Train: April + May. Then a 7-day gap. Test: June (until the cutoff above).
train = df[df["date"] <= "2026-05-31"]
test  = df[df["date"] >= "2026-06-07"]      # do not touch, do not resample

# ---------- 3. DOWNSAMPLE HEALTHY DRIVES (TRAIN ONLY) ----------
# "failed" = drives that have at least one positive row in train
failed  = set(train.loc[train["target"] == 1, "serial_number"])

# all other drives in train are "healthy"
is_failed = train["serial_number"].isin(failed)
healthy   = np.array(train.loc[~is_failed, "serial_number"].unique())

HEALTHY_PER_FAILED = 1        # 1 healthy drive per failed drive (try 10 later)
n_keep = min(len(failed) * HEALTHY_PER_FAILED, len(healthy))

rng  = np.random.default_rng(0)
keep = set(rng.choice(healthy, size=n_keep, replace=False))

# keep every failed drive + the randomly chosen healthy drives (whole histories)
train_small = train[train["serial_number"].isin(failed | keep)].copy()

# ---------- 4. DROP COLUMNS THAT CONTAIN THE ANSWER ----------
# These reveal the label, so the model would cheat if it saw them.
for d in (train_small, test):
    d.drop(columns=["fail_date", "days_to_fail", "failure"], inplace=True)

# ---------- 5. CHECK ----------
print("train_small rows:", len(train_small))
print(train_small["target"].value_counts())
print("test rows:", len(test), " positives in test:", test["target"].sum())
```

```python
# Run these cells AFTER your label cell (the one that builds `target`
# and filters df to date <= max - 7 days). Each "# %%" block is one cell.
#
# Needs in memory: df with columns
#   date, serial_number (category), model (category), capacity_bytes,
#   smart_5/9/187/188/197/198_raw, target
# and df sorted by serial_number, date (you already did this).

# %% CELL 1: settings
import numpy as np
import pandas as pd

TOP_K = 14          # how many features to keep after selection
P_TEST = 0.03       # fraction of healthy test drives we score (speed)
SVM_CAP = 40000     # max training rows for the SVM (RBF SVM is slow)
SEED = 0
rng = np.random.default_rng(SEED)

```

```python
# %% CELL 2: feature engineering
# Idea: a single reading says "how bad is it now".
# A change over time says "is it getting worse". The paper only used
# single readings. We add changes over 7 and 14 days.

COUNTERS = ["smart_5_raw", "smart_187_raw", "smart_197_raw", "smart_198_raw"]
ALL_CNT = COUNTERS + ["smart_188_raw"]


def signed_log(x):
    # compress huge values but keep the sign (a counter can go down)
    return np.sign(x) * np.log1p(np.abs(x))


def vendor_of(model):
    m = str(model).upper()
    if m.startswith("ST") or m.startswith("SEAGATE"):
        return "SEAGATE"
    if m.startswith("WDC") or m.startswith("WD"):
        return "WDC"
    if m.startswith("TOSHIBA") or m.startswith("MG") or m.startswith("MD"):
        return "TOSHIBA"
    if m.startswith("HGST") or m.startswith("HUH") or m.startswith("HMS"):
        return "HGST"
    return "OTHER"


def add_features(df):
    # df MUST be sorted by serial_number, then date.
    n = len(df)
    codes = df["serial_number"].cat.codes.to_numpy()
    days = df["date"].to_numpy().astype("datetime64[D]").astype("int64")
    out = {}

    # current level, log scale. NaN stays NaN (drive does not report it)
    for c in ALL_CNT:
        x = df[c].to_numpy(dtype="float32")
        out[c.replace("_raw", "") + "_log"] = np.log1p(np.clip(x, 0, None))

    # change over k days. Only valid if the row k rows back is the
    # same drive AND exactly k calendar days earlier (handles missing days).
    for c in COUNTERS:
        x = df[c].to_numpy(dtype="float32")
        for k in (7, 14):
            lag = np.full(n, np.nan, dtype="float32")
            lag[k:] = x[:-k]
            ok = np.zeros(n, dtype=bool)
            ok[k:] = (codes[k:] == codes[:-k]) & (days[k:] - days[:-k] == k)
            d = np.where(ok, x - lag, np.nan).astype("float32")
            out[c.replace("_raw", "") + f"_d{k}"] = signed_log(d)

    out["power_on_days"] = df["smart_9_raw"].to_numpy(dtype="float32") / 24
    out["capacity_tb"] = (df["capacity_bytes"].to_numpy(dtype="float64") / 1e12).astype("float32")

    # some vendors never report 187 or 188. "Missing" itself is information.
    out["s187_missing"] = df["smart_187_raw"].isna().to_numpy().astype("int8")
    out["s188_missing"] = df["smart_188_raw"].isna().to_numpy().astype("int8")

    # vendor one-hot (map per model category, then index: fast)
    cats = df["model"].cat.categories
    vendor_by_cat = np.array([vendor_of(m) for m in cats])
    vendors = vendor_by_cat[df["model"].cat.codes.to_numpy()]
    for v in np.unique(vendors):
        out[f"vendor_{v}"] = (vendors == v).astype("int8")

    return pd.concat([df, pd.DataFrame(out, index=df.index)], axis=1)


df = df.drop(columns=["fail_date", "days_to_fail", "failure"], errors="ignore")
df = add_features(df)

NEW = [c for c in df.columns if c.endswith(("_log", "_d7", "_d14"))
       or c in ("power_on_days", "capacity_tb", "s187_missing", "s188_missing")
       or c.startswith("vendor_")]
FEATS = NEW
print(len(FEATS), "features")
print(df[[c for c in FEATS if c.startswith("vendor_")]].sum())  # check no big "OTHER"
```

```python


# %% CELL 3: split (same logic as before, now with features)
train = df[df["date"] <= "2026-05-31"]
test = df[df["date"] >= "2026-06-07"]

# --- train: all failed drives + equal number of random healthy drives
failed = set(train.loc[train["target"] == 1, "serial_number"])
is_failed = train["serial_number"].isin(failed)
healthy = np.array(train.loc[~is_failed, "serial_number"].unique())
n_keep = min(len(failed), len(healthy))
keep = set(rng.choice(healthy, size=n_keep, replace=False))
train_small = train[train["serial_number"].isin(failed | keep)].copy()

# --- test: keep ALL failing drives, sample P_TEST of healthy drives,
# and give sampled healthy rows weight 1/P_TEST so metrics stay realistic.
fail_test = set(test.loc[test["target"] == 1, "serial_number"])
is_ft = test["serial_number"].isin(fail_test)
healthy_t = np.array(test.loc[~is_ft, "serial_number"].unique())
keep_t = set(rng.choice(healthy_t, size=int(len(healthy_t) * P_TEST), replace=False))
test_s = test[is_ft | test["serial_number"].isin(keep_t)].copy()
test_s["w"] = np.where(test_s["serial_number"].isin(fail_test), 1.0, 1.0 / P_TEST)

N_DAYS = test["date"].nunique()
FLEET_PER_DAY = len(test) / N_DAYS
print("train_small:", train_small.shape, "pos:", int(train_small["target"].sum()))
print("test_s:", test_s.shape, "real positives:", int(test_s["target"].sum()))
print(f"test days: {N_DAYS}, fleet size per day: {FLEET_PER_DAY:,.0f}")
```

```python

# %% CELL 4: feature selection (course topic)
# Two views of "which features matter":
#  - Spearman correlation (what the paper used)
#  - Mutual information (catches non-linear links too)
from sklearn.feature_selection import mutual_info_classif

med = train_small[FEATS].median()          # fill NaN with train medians
Xtr_all = train_small[FEATS].fillna(med)
ytr_all = train_small["target"].to_numpy()

sub_idx = rng.choice(len(Xtr_all), size=min(30000, len(Xtr_all)), replace=False)
mi = mutual_info_classif(Xtr_all.iloc[sub_idx], ytr_all[sub_idx], random_state=SEED)
sp = Xtr_all.corrwith(pd.Series(ytr_all, index=Xtr_all.index), method="spearman")

rank = (pd.DataFrame({"mutual_info": mi, "spearman": sp}, index=FEATS)
        .sort_values("mutual_info", ascending=False))
print(rank.round(4))

selected = list(rank.index[:TOP_K])
print("\nselected:", selected)
```

```python

# %% CELL 5: models (course topics: kNN and SVM)
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split

Xtr = train_small[selected].fillna(med[selected])
ytr = train_small["target"].to_numpy()
Xte = test_s[selected].fillna(med[selected])
yte = test_s["target"].to_numpy()
w = test_s["w"].to_numpy()

# kNN and SVM care about feature scale, so StandardScaler goes first.
knn = make_pipeline(StandardScaler(),
                    KNeighborsClassifier(n_neighbors=50, weights="distance", n_jobs=-1))
svm = make_pipeline(StandardScaler(),
                    SVC(kernel="rbf", C=1.0, gamma="scale", class_weight="balanced"))

knn.fit(Xtr, ytr)

Xs, _, ys, _ = train_test_split(Xtr, ytr, train_size=min(SVM_CAP, len(Xtr)),
                                stratify=ytr, random_state=SEED)
svm.fit(Xs, ys)

scores = {
    "kNN": knn.predict_proba(Xte)[:, 1],
    "SVM": svm.decision_function(Xte),
}

# Simple rule baseline: sum of the four error counters (log scale).
# If a model cannot beat this, it is not adding value.
base_cols = [c.replace("_raw", "") + "_log" for c in COUNTERS]
scores["Rule (sum of 4 counters)"] = test_s[base_cols].fillna(0).sum(axis=1).to_numpy()
```

```python

# %% CELL 6: realistic evaluation
from sklearn.metrics import average_precision_score, roc_auc_score


def evaluate(s, y, w, n_days, budgets=(10, 50, 200, 1000)):
    res = {
        "PR_AUC": average_precision_score(y, s, sample_weight=w),
        "ROC_AUC": roc_auc_score(y, s, sample_weight=w),
    }
    order = np.argsort(-s, kind="stable")
    ws, ys_ = w[order], y[order]
    cum_all = np.cumsum(ws)
    cum_pos = np.cumsum(ws * ys_)
    total_pos = cum_pos[-1]
    for b in budgets:
        i = np.searchsorted(cum_all, b * n_days, side="right")  # rows inside budget
        res[f"recall@{b}/day"] = cum_pos[i - 1] / total_pos if i > 0 else 0.0
        res[f"prec@{b}/day"] = cum_pos[i - 1] / cum_all[i - 1] if i > 0 else np.nan
    return res


table = pd.DataFrame({k: evaluate(v, yte, w, N_DAYS) for k, v in scores.items()}).T
print(f"Random-guess PR_AUC would be {np.average(yte, weights=w):.5f}")
print(table.round(4).to_string())
```
